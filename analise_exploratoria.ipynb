## Visualização dos dados com Matplotlib e Seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Estilo visual do gráfico
sns.set_theme(style="whitegrid")

# tamanho da figura
plt.figure(figsize=(12, 6))

# Boxplot de comparação
grafico = sns.boxplot(
    data=df_series, 
    x="Streaming/Canal", 
    y="Nota TVMaze", 
    hue="Streaming/Canal",
    palette="viridis",
    legend=False
)

# Ajusta os rótulos e o título
plt.title("Distribuição de Notas das Séries por Plataforma", fontsize=14, pad=15)
plt.xlabel("Serviço de Streaming / Canal", fontsize=12)
plt.ylabel("Nota (TVMaze)", fontsize=12)

# Virando as caixas do plot para não ficarem sem respiro no gráfico
plt.xticks(rotation=45, ha='right')

# Exibe o gráfico
plt.tight_layout()

plt.savefig("boxplot_series.png", dpi=300, bbox_inches='tight')
plt.show()
```
